## **DISCIPLINE**

## INVARIANT: GIVING UP FREEDOM SO THAT COPMPLEXITY DOESN'T EXPLODE

## What is discipline?
- The act of intentionally restricting your design choices so that you can work more productively at a higher level of abstraction 

- An example of interchangable parts is a familiar application of disciple. 

- When we build a computer - why is it so easy to swap out parts? It is because of discipline - we have a STANDARDIZED set of components that work with each other with WELL-DEFINED TOLERANCES 

- This has to benefit of abstraction - we don't need to know how a CPU is built whenever we are assembling a computer

## How does this relate to digital systems?
- Digital circuits use discrete voltages, whereas analog circuits use continuous voltages - in other words digital circuits are a subset of analog circuits - less range than an analog circuit. 

- this simplicity is what allows us to design them more quickly - limiting in this instance - the actf of discipline is being able to limit the bounds of what we are doing in order to remove the complexity of building a circuit 

# Why Digital is outperforming analog 

- By combining components into sophisticated systems that outperform analog

- Analog systems lose because errors accumulate continuously and there’s no natural reset, so noise, drift, and imprecision compound as systems scale.\

- Digital systems introduce hard thresholds that erase small errors at every step, letting signals be regenerated cleanly through long chains of logic.

- By throwing away fine-grained detail, digital gains reliability, verifiability, and scalability, which is why large computation is possible.

## The Three -Y's: Hierarchy, Modularity, Regularity 

- Hierarchy means to take a system and divide it into modules, then further subdivide - a recursive way to solve a problem basically

- Modularity means that these modules have easy ways to interact with each other

- Regularity is how to connect/reuse those modules

An example of this using legos: we have a lego doggie we bought from ebay, pre-built for us. We want to make it into something else. We must break that dog into a **hierachy** of specific pieces.

Every lego has little studs on the top of the lego brick. This is the concept of **modularity** - the very idea of having studs on the top of the lego means that this lego brick can connect to other lego pieces without too many consequences. 

We can connect these lego pieces to other lego pieces. This act of connecting those lego pieces is what we call the act of **Regularity**. The unification of a lego (module) - this reduces the number of new pieces we have to make. 


INVARIANTS:
1. Hierarchy: 
    - each component can be further subdivided into smaller components
2. Modularity: 
    - each component should have a well-defined function and interface
        - function: what does it do?
        - interface: how does one access it?
    - assuming those conditions are met and matching with other components one can change out the pieces in a lego without consequences
3.  Regularity: 
    - Interchangable parts are a good idea - it allows for whatever is broken to be changed out without much hassle

## **DECIMAL-TO-DECIMAL**

```python
# Challenge: decompose a number into place values

# x = input("Enter a number")
x = '9432'
try:
    digits = []
    for ch in x:
        digits.append(int(ch))

    print(digits)

    total = 0
    terms = []

    for i in range(len(digits)):
        power = (len(digits) - i - 1)
        terms.append(f"({digits[i]} * 10^{power})")
        total += (digits[i] * (10 ** power))

    print(" + ".join(terms))
    print("Sum:", total)

except Exception as e:
    print("Error:", e)
```

## **BINARY-TO-DECIMAL**

```python
# BINARY TO DECIMAL

# x = input("Enter a binary value")
x = '01011100'

def binary_to_decimal(x):

    sum_ = 0
    for i in range(len(x)-1, -1, -1):
        power = len(x) - i - 1
        # print(2, "^", power," * ", x[i])
        sum_ += 2 ** power * int(x[i])
    return sum_

print(binary_to_decimal(x))
```

## **DECIMAL-TO-BINARY**

```python
import math

def decimal_to_binary(x):
    result = ""
    while not (x == 0): 
        remainder = x % 2
        x = math.floor(x / 2)
        result += str(remainder)
    return result[::-1]

x = decimal_to_binary(92)
print(x)
```

#### It may be useful to remember the first 16 powers of binary values

```python
for i in range(16):
    print(decimal_to_binary(2**i), binary_to_decimal(decimal_to_binary(2**i)))
```

## BYTES, NIBBLES, AND WORDS

### Bits, Nibbles, and Bytes
- A **bit** is a single binary digit: `0` or `1`
- A **nibble** is **4 bits**
- A **byte** is **8 bits**

**Invariant:**  
2 nibbles = 1 byte

---

### Bytes and Combinations
- A byte has:
  2^8 = 256 possible bit patterns

**Example:**
[ 0 1 1 1 ] [ 0 1 0 1 ]


- These 256 patterns represent values:
  0 to 255  
  (because counting starts at 0)

---

### Words
- A **word** is a fixed-size chunk of data used by a microprocessor
- **Word size depends on the architecture**, for example:
  - 16-bit
  - 32-bit
  - 64-bit

---

## SIGNIFICANCE

### Bit-Level Significance
- The **Least Significant Bit (LSB)** is the bit in the 2^0 (1’s) position
- The **Most Significant Bit (MSB)** is the bit with the highest power of 2

---

### Byte-Level Significance (Within a Word)
- In a multi-byte word:
  - The **Least Significant Byte (LSByte)** contains the lowest-weight bits
  - The **Most Significant Byte (MSByte)** contains the highest-weight bits

---

### Core Invariant
- Significance is determined by **power of two**, not by storage order
